```python
import pandas as pd

#LOAD DATA
df = pd.read_csv("netflix_titles.csv")

# Count Records,Movies and TV Shows BEFORE cleaning
total_records = df.shape[0]
total_movies = df[df['type'] == 'Movie'].shape[0]
total_tvshows = df[df['type'] == 'TV Show'].shape[0]
print("Total Records:", total_records)
print("Total Movies:", total_movies)
print("Total TV Shows:", total_tvshows)
```

```python
#BASIC DATA OVERVIEW 
print("HEAD")
print(df.head())

print("\nINFO")
print(df.info())

print("\nDESCRIBE")
print(df.describe(include='all'))

print("\nDataset Shape:", df.shape)
print("\nColumns:", df.columns.tolist())

print("\nCOUNT")
print(df.count())

print("\nNULL VALUES BEFORE CLEANING")
print(df.isnull().sum())

print("\nDIRECTOR VALUE COUNTS")
print(df["director"].value_counts())

print("\nCAST VALUE COUNTS")
print(df["cast"].value_counts())

print("\nCOUNTRY VALUE COUNTS")
print(df["country"].value_counts())

```

```python
#HANDLE MISSING VALUES
df['director'] = df['director'].fillna('Unknown')
df['cast'] = df['cast'].fillna('Unknown')
df['country'] = df['country'].fillna('Unknown')
df['date_added'] = df['date_added'].fillna('Unknown')
df['rating'] = df['rating'].fillna('Unknown')
df['duration'] = df['duration'].fillna("Unknown")

print("\nNULL VALUES AFTER BASIC FILLING")
print(df.isnull().sum())
```

```python
# CLEAN & EXTRACT DURATION
df['duration'] = df['duration'].str.replace("Seasons", "Season")

df['duration_num'] = df['duration'].str.extract(r'(\d+)').astype(float)
df['duration_type'] = df['duration'].str.extract(r'(min|Season)')

print("\nNULL VALUES AFTER DURATION EXTRACTION")
print(df.isnull().sum())

df.drop(columns=['duration'], inplace=True)
```

```python
#REMOVE EXTRA WHITESPACE
df['title'] = df['title'].str.strip()
df['director'] = df['director'].str.strip()
df['cast'] = df['cast'].str.strip()
df['country'] = df['country'].str.strip()
df['rating'] = df['rating'].str.strip()


#REMOVE DUPLICATES
df.drop_duplicates(inplace=True)

print("\nNULL VALUES AFTER REMOVING DUPLICATES")
print(df.isnull().sum())
```

```python
#HANDLE REMAINING NULL VALUES
df['duration_num'] = df['duration_num'].fillna(0)
df['duration_type'] = df['duration_type'].fillna('Unknown')

print("\nFINAL NULL VALUES")
print(df.isnull().sum())

print("\nFINAL DATA PREVIEW")
print(df.head())
```

```python
# SAVE CLEANED DATASET
df.to_csv("netflix_cleaned_new3.csv", index=False)

print("\nCleaning Completed — File saved as netflix_cleaned_new3.csv")
```

```python
import pandas as pd

# Load cleaned dataset
df = pd.read_csv("/Workspace/Users/hemavarshinis.23aid@kongu.edu/netflix_cleaned_new3.csv")

#TOP 10 COUNTRIES
top_10_countries = (
    df['country']
    .value_counts()
    .head(10)
)
print("TOP 10 COUNTRIES:\n", top_10_countries, "\n")


#TOP 10 GENRES
top_10_genres = (
    df['listed_in']
    .str.split(", ")
    .explode()
    .value_counts()
    .head(10)
)
print("TOP 10 GENRES:\n", top_10_genres, "\n")


#TOP 5 ACTORS
top_5_actors = (
    df['cast']
    .str.split(", ")
    .explode()
    .value_counts()
    .head(5)
)
print("TOP 5 ACTORS:\n", top_5_actors, "\n")


#TOP 5 DIRECTORS
top_5_directors = (
    df['director']
    .str.split(", ")
    .explode()
    .value_counts()
    .head(5)
)
print("TOP 5 DIRECTORS:\n", top_5_directors, "\n")


#CONTENT ADDED OVER YEARS
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')

df['year_added'] = df['date_added'].dt.year

content_by_year = (
    df.groupby('year_added')['show_id']
      .count()
      .dropna()
)
print("CONTENT ADDED OVER YEARS:\n", content_by_year, "\n")


#MOVIE DURATION DISTRIBUTION
top_10_duration = (
    df['duration_num']
    .value_counts()
    .head(10)
)
print("TOP 10 OVERALL DURATIONS:\n", top_10_duration)
```
