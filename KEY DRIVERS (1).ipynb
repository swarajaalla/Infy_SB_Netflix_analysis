NUMBER OF TITLES RELEASED TRENDS

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv('netflix_updated.csv')

# --- Data Processing ---
# Count the number of titles released each year and sort by year
production_trend = df['release_year'].value_counts().sort_index()

# Filter for years from 2000 onwards for a clearer plot trend
production_trend_filtered = production_trend[production_trend.index >= 2000]


# --- Plotting ---
plt.figure(figsize=(12, 6))

# Create the line plot
production_trend_filtered.plot(kind='line', marker='o', color='skyblue')

# Set the title and labels
plt.title('Netflix Content Release Trend (Titles Released Per Year)', fontsize=16)
plt.xlabel('Release Year', fontsize=12)
plt.ylabel('Number of Titles Released', fontsize=12)

# Customizing x-axis ticks for better readability
years = production_trend_filtered.index.tolist()
# Only show every 2nd year on the x-axis if there are many years
tick_step = 2 if len(years) > 20 else 1
plt.xticks(years[::tick_step], rotation=45)

plt.grid(axis='y', linestyle='--', alpha=0.7)

# Highlight the year with the maximum number of releases
max_year = production_trend_filtered.idxmax()
max_count = production_trend_filtered.max()
plt.axvline(x=max_year, color='red', linestyle='--', linewidth=1, alpha=0.7)
plt.text(max_year, max_count * 0.9, f'Peak: {max_year}', color='red', ha='right', va='center')

plt.tight_layout()

# Save the plot to a file
plt.savefig('netflix_production_trend_yearly.png')
# plt.show() # Use this command if you are running in a local environment (not a notebook)
print("Plot saved as 'netflix_production_trend_yearly.png'")
```

TOP 10 PROLIFIC DIRECTOR

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
try:
    df = pd.read_csv('netflix_updated.csv')
except FileNotFoundError:
    print("Error: netflix_updated.csv not found.")
    # Exit or handle the error appropriately
    raise

# --- Data Preparation ---
# Replace 'Unknown' with NaN for proper dropping/counting, and drop rows where 'director' is missing
df_directors = df.dropna(subset=['director'])
df_directors = df_directors[df_directors['director'] != 'Unknown'].copy()

# Function to expand comma-separated entries into individual counts
def count_frequent_items(series):
    # Split the string by comma and strip whitespace, then count occurrences
    all_items = series.str.split(',\s*').explode()
    return all_items.value_counts()

# Get the counts for directors
director_counts = count_frequent_items(df_directors['director'])

# Get the Top 10 most frequent directors
top_10_directors = director_counts.head(10)

# --- Visualization (Bar Plot) ---
plt.figure(figsize=(10, 7))

# Use seaborn for a visually appealing bar chart
sns.barplot(x=top_10_directors.values, y=top_10_directors.index, palette='Spectral')

# Customizing the plot
plt.title('Top 10 Most Prolific Netflix Directors', fontsize=16)
plt.xlabel('Number of Titles Directed', fontsize=12)
plt.ylabel('Director Name', fontsize=12)

plt.tight_layout()

# Save or display the plot
plt.savefig('netflix_top_10_directors.png')
print("Bar plot code executed. Plot saved as 'netflix_top_10_directors.png'")
```

IMPORTANT GENRES

```python
import pandas as pd

# Load the dataset
try:
    df = pd.read_csv('netflix_updated.csv')
except FileNotFoundError:
    print("Error: netflix_updated.csv not found.")
    raise

# --- 1. Data Cleaning and Preparation ---
df_clean = df[(df['country'] != 'Unknown') & (df['country'].notna())].copy()
df_clean['primary_country'] = df_clean['country'].apply(lambda x: x.split(',')[0].strip())

# --- 2. Explode Genres for Counting (FIXED LINE) ---
# Use r'...' for the regular expression to treat '\s' as a literal backslash followed by 's',
# which is what the regex engine needs for matching whitespace.
df_exploded = df_clean.assign(listed_in=df_clean['listed_in'].str.split(r',\s*')).explode('listed_in')

# --- 3. Group and Count ---
genre_country_counts = df_exploded.groupby(['primary_country', 'listed_in']).size().reset_index(name='count')

# --- 4. Find the Most Important (Most Frequent) Genre per Country ---
genre_country_counts_sorted = genre_country_counts.sort_values(
    by=['primary_country', 'count'],
    ascending=[True, False]
)
most_important_genre = genre_country_counts_sorted.drop_duplicates(
    subset=['primary_country'],
    keep='first'
)

# --- 5. Final Table Formatting ---
final_table = most_important_genre.sort_values(by='count', ascending=False).head(10)
final_table = final_table[['primary_country', 'listed_in', 'count']]
final_table.columns = ['Country', 'Most Important Genre', 'Count of Top Genre']

# Print the final result using to_string()
print(final_table.to_string(index=False))
```

COUNT OF TOP GENRES BY COUNTRY

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
try:
    df = pd.read_csv('netflix_updated.csv')
except FileNotFoundError:
    print("Error: netflix_updated.csv not found.")
    raise

# --- 1. Data Cleaning and Preparation ---
# Filter out rows where 'country' is missing or 'Unknown'
df_clean = df[(df['country'] != 'Unknown') & (df['country'].notna())].copy()
# Extract only the primary country (the first one listed)
df_clean['primary_country'] = df_clean['country'].apply(lambda x: x.split(',')[0].strip())

# --- 2. Explode Genres for Counting ---
# Split the 'listed_in' column and expand the DataFrame.
# Using r',\s*' for regex splitting to prevent the SyntaxWarning.
df_exploded = df_clean.assign(listed_in=df_clean['listed_in'].str.split(r',\s*')).explode('listed_in')

# --- 3. Group and Count ---
# Group by primary country and genre, and count the occurrences
genre_country_counts = df_exploded.groupby(['primary_country', 'listed_in']).size().reset_index(name='Count')

# --- 4. Find the Highest Count Genre per Country ---
# Sort the counts in descending order within each country group
genre_country_counts_sorted = genre_country_counts.sort_values(
    by=['primary_country', 'Count'],
    ascending=[True, False]
)

# Keep only the top row (the highest count genre) for each country
most_important_genre = genre_country_counts_sorted.drop_duplicates(
    subset=['primary_country'],
    keep='first'
)

# --- 5. Prepare Data for Plotting ---
# Rename columns
final_table = most_important_genre[['primary_country', 'listed_in', 'Count']].copy()
final_table.columns = ['Country', 'Most Frequent Genre', 'Count']

# Sort the final table by the count (descending) and get the top 10
top_10_plot = final_table.sort_values(by='Count', ascending=False).head(10)


# --- 6. Visualization (Vertical Bar Plot) ---
plt.figure(figsize=(12, 7))

# Create a vertical bar plot
sns.barplot(
    x='Country',
    y='Count',
    data=top_10_plot,
    hue='Most Frequent Genre', # Color bars based on the genre
    dodge=False, # Ensure one bar per country
    palette='Spectral'
)

# Customizing the plot
plt.title('Top 10 Countries by Count of Most Frequent Genre', fontsize=16)
plt.xlabel('Country', fontsize=12)
plt.ylabel('Count of Most Frequent Genre', fontsize=12)

# Rotate x-axis labels for readability
plt.xticks(rotation=45, ha='right')

# Add legend outside the plot area
plt.legend(title='Most Frequent Genre', bbox_to_anchor=(1.05, 1), loc='upper left')

plt.tight_layout(rect=[0, 0, 0.85, 1]) # Adjust layout to ensure the legend fits

# Save the plot
plt.savefig('top_genre_count_by_country_vertical_plot.png')
print("Plot code executed. Plot saved as 'top_genre_count_by_country_vertical_plot.png'")
```

CONTENT AVAILABILITY

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
file_name = 'netflix_updated.csv'
df = pd.read_csv(file_name)

# 1. Data Cleaning and Pre-processing
df['country'] = df['country'].replace('Unknown', np.nan)
df['listed_in'] = df['listed_in'].replace('Unknown', np.nan)

# 2. Explode the Country column (one row per country)
country_df = df.assign(country=df['country'].str.split(', ')).explode('country').copy()

# 3. Explode the Genre (listed_in) column on the exploded country data
country_genre_df = country_df.assign(listed_in=country_df['listed_in'].str.split(', ')).explode('listed_in').copy()

# Drop rows where BOTH country and genre are NaN (or were 'Unknown' originally)
country_genre_df.dropna(subset=['country', 'listed_in'], inplace=True)

# 4. Calculate Country-Genre Availability Counts
country_genre_counts = country_genre_df.groupby(['country', 'listed_in']).size().reset_index(name='Content_Count')

# 5. Filter for Top Content Drivers
top_countries = country_genre_df['country'].value_counts().nlargest(10).index.tolist()
top_genres = country_genre_df['listed_in'].value_counts().nlargest(10).index.tolist()

filtered_counts = country_genre_counts[
    country_genre_counts['country'].isin(top_countries) &
    country_genre_counts['listed_in'].isin(top_genres)
]

# 6. Pivot the filtered data for Heatmap visualization
pivot_table = filtered_counts.pivot_table(
    index='country',
    columns='listed_in',
    values='Content_Count',
    fill_value=0
)

# 7. Visualization (Heatmap)
plt.figure(figsize=(12, 10))
sns.heatmap(
    pivot_table,
    annot=True,         # Show the count in each cell
    fmt='.0f',          # CORRECTED: Use .0f to format as integer-like string to avoid ValueError
    cmap='YlGnBu',
    linewidths=0.5,
    linecolor='black',
    cbar_kws={'label': 'Number of Titles'}
)

plt.title('Content Availability: Top 10 Countries vs. Top 10 Genres', fontsize=16)
plt.xlabel('Genre', fontsize=12)
plt.ylabel('Country', fontsize=12)
plt.yticks(rotation=0)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()

plt.show()
plt.close()
```

PERCENTAGE OF  MOVIE AND TV SHOW PERCENTAGE

```python
import pandas as pd

# Load the dataset
try:
    df = pd.read_csv('netflix_updated.csv')
except FileNotFoundError:
    print("Error: netflix_updated.csv not found.")
    raise

# --- 1. Data Cleaning and Preparation ---
# Filter out rows where 'country' is missing or 'Unknown'
df_clean = df[(df['country'] != 'Unknown') & (df['country'].notna())].copy()

# Extract only the primary country (the first one listed)
df_clean['primary_country'] = df_clean['country'].apply(lambda x: x.split(',')[0].strip())

# --- 2. Group, Count, and Unstack ---
# Group by primary_country and content type, and count the occurrences.
# unstack() pivots the 'type' categories (Movie, TV Show) from rows into separate columns.
country_type_counts = df_clean.groupby(['primary_country', 'type']).size().unstack(fill_value=0)

# Rename columns (assuming 'Movie' and 'TV Show' are the column names after unstacking)
country_type_counts.columns = ['Movie Count', 'TV Show Count']

# --- 3. Calculate Percentages ---
# Calculate the total count
country_type_counts['Total Titles'] = country_type_counts['Movie Count'] + country_type_counts['TV Show Count']

# Calculate percentages
country_type_counts['% Movie'] = (country_type_counts['Movie Count'] / country_type_counts['Total Titles']) * 100
country_type_counts['% TV Show'] = (country_type_counts['TV Show Count'] / country_type_counts['Total Titles']) * 100

# --- 4. Final Formatting and Sorting ---
# Sort by Total Titles in descending order
final_table = country_type_counts.sort_values(by='Total Titles', ascending=False)

# Select the required columns and reset index
final_table = final_table[['Movie Count', '% Movie', 'TV Show Count', '% TV Show']].reset_index()
final_table.columns = ['Country', 'Movie Count', '% Movie', 'TV Show Count', '% TV Show']

# Format percentages to one decimal place and append '%'
final_table['% Movie'] = final_table['% Movie'].round(1).astype(str) + '%'
final_table['% TV Show'] = final_table['% TV Show'].round(1).astype(str) + '%'

# Print the final result using to_string()
print(final_table.to_string(index=False))
```
